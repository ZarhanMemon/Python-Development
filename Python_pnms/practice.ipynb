```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Beginner
# - Load the tips dataset and visualize the distribution of total_bill values.
# - Create a scatter plot showing the relationship between total_bill and tip.
# - Plot a bar chart of average tips grouped by day of the week.
# - Show the count of male vs. female customers using a countplot.

tips = sns.load_dataset("tips")

tips.head()
# total_bills = tips['total_bill']

# total_bills
```

```python
sns.scatterplot( data= tips , x='total_bill' , y='tip')

plt.title("Sctater bills v/s tip")
plt.show()
```

```python

ax = sns.barplot( data= tips , y='tip' , x='day' )

# Iterate over all containers and add labels
for container in ax.containers:
    ax.bar_label(container)

plt.show()
```

```python

ax = sns.countplot( data= tips , x='sex' , hue='sex')

# Iterate over all containers and add labels
for container in ax.containers:
    ax.bar_label(container)

plt.show()
```

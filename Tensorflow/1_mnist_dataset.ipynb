# Understanding and visualizing MNIST Dataset
The MNIST dataset is like the "Hello World" of deep learning. It contains 70,000 grayscale images of handwritten digits (0–9), each of size 28x28 pixels. Before we train a neural network, it’s important to understand what we’re working with.

Today we'll:

- Load the MNIST dataset using TensorFlow
- Visualize a few digit samples
- Understand the data format

----
<br>

## 1. Loading the MNIST data of 0-9 number images :

```python
import tensorflow as tf
import matplotlib.pyplot as plt
```

```python

# Loading the MNIST data from tf
mnist_data = tf.keras.datasets.mnist.load_data()

( x_test, y_test ) , ( x_train , y_train ) = mnist_data



#--------------------------------------------


# Shape of data
print(f" Testing data shape : { x_test.shape}")
print(f" Training data shape : { x_train.shape}")
```

- We have 60,000 training images and 10,000 test images.
- Each image is 28x28 pixels.
- Each label is a number from 0 to 9.

------
<br>

## 2. Exploring the data:

```python

plt.figure(figsize=(10, 2))   # changing figure/img size


for i in range(5):           # print the 5 img in loop 
    
    plt.subplot(1, 5, i + 1)      #making the subplot ( row = 1 , col = 5 , img)
    
    plt.title(y_test[i])                   # show the x_test img ->  label in y_test
    plt.imshow( x_test[i] , cmap="gray")   # showing img in x_test , in the gray-scale histogram theme
    
    plt.axis("off")   
    
    
plt.tight_layout()
```

<br>

### What You Should Notice
1. The digits vary in writing style, which makes this dataset great for teaching computers to generalize.
2. All images are normalized 28x28 grayscale—no color channels.
3. The label is not embedded in the image; it's provided separately.

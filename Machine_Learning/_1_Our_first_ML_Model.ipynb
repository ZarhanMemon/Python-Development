Step 2: Make Your First Machine Learning Model (in 5 minutes!)

Here’s a "Can we predict if a fruit is an apple or an orange?" example:

The idea:
We have some fruits.
Apples are heavy and red
Oranges are lighter and orange-colored

At the first glance, it seems easy to tell them apart, right? You can always compare Orange pixels and red pixels, or check the weight. But as you keep getting more fruits, it becomes harder to write rules for every single fruit and with the messy real world, rules you write starts to break. Let’s train a machine to learn this!

```python

# Step 1: Importing the Decision Tree Classifier - because we want to use it for our classification task of Orange or Apple
from sklearn import tree


#==================================


# Step 2: Creating the dataset - we will create a simple dataset with two features (color and size) and a target variable (fruit type)

feature = [ 
             [120, 0],  # [ weight(g) , color (0 for orange, 1 for red) ]
             [180, 1],  # apple
             [170, 0],  # orange
             [130, 1],  # apple
             [120, 1],  # apple
]

label = [ 'orange', 'apple', 'orange', 'apple', 'apple' ]  # Target variable



#=====================================


# Step 3: Training the Decision Tree Classifier - we will fit the model to our dataset

clf = tree.DecisionTreeClassifier()  # Create an instance of the Decision Tree Classifier

clf = clf.fit( feature  , label)     # Fit/loading the model to the dataset (training)


#=====================================

# Step 4: Making Predictions - we will use the trained model to make predictions on new data

clf.predict([
             [150, 0],
             [130, 0],
             [160, 1]
             ])  # Predicting by inputing the [ [weight, color] , [..],,] of the fruit we want to classify
```

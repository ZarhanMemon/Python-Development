Measuring Accuracy of our Predictions


- When we build a machine learning model, especially a classification model (which predicts categories like "spam" or "not spam", "dog" or "cat"), it's important to measure how well the model is performing.

- One of the most basic ways to evaluate a classification model is accuracy.

-----------

.

1. What is Accuracy?



- In simple terms, accuracy tells us how often the model was right.

- If you gave your model 100 questions to answer, and it got 90 of them correct, its accuracy would be:


```python
Accuracy = Correct Predictions / Total Predictions = 90 / 100 = 90%

# So, accuracy is just the fraction of predictions the model got right.
```

```python
%pip install openpyxl

from sklearn.ensemble import RandomForestClassifier   # accuracy = 100.0 
from sklearn.tree import DecisionTreeClassifier      # accuracy = 100.0 
from sklearn.naive_bayes import GaussianNB           # accuracy = 97.05882352941177

import pandas as pd            
```

```python
data = pd.read_excel('train_IrisFlowerDataset.xlsx')     # or u can also use the csv file
```

```python

print(data.head())     # to see the first 5 rows of the dataset


features = data.drop( columns=['class'])     # the features are all the columns except the 'class' column

labels = data['class']                       # the labels are the 'class' column which contains the type of flower , 
                                             # which is the target variable we want to predict


# OR  features = data.iloc[: , :-1]     # removing the last column "class"
#     labels = data.iloc[: , -1:-1]        # removing all cols except the last column "class"


print(features.head()  , '\n\n' , labels.head())     
```

```python

# Different models to choose from 
# - we get different accuracy for different models  

# model = RandomForestClassifier()
model = DecisionTreeClassifier()
# model = GaussianNB()


# TRAINING THE MODEL - by train data
model = model.fit(features , labels)     # fit() means find patterns in the data
```

```python

test_data = pd.read_excel('test_IrisFlowerDataset.xlsx')     # for testing model on new data 


p_features = test_data.drop( columns=['class'])     # these are the testing feature data which we want to predict the labels for
# print(p_features.head())     


# ------------------------


predict_1 = model.predict([[5.1, 3.5, 1.4, 0.2]])    # -> class = Iris-setosa

predict_labels = model.predict(p_features)


# ------------------------

# It predict the name/class of flower , by using the train data exp

# print(predict_1)       

print(predict_labels)         
```

-----------

2. Evaluating Accuracy



- Accuracy = (Correct Predictions) / (Total Predictions)



This is the simplest and most intuitive way to check if the model is doing a good job.

```python
# Answer of the test data
correct_labels = test_data['class']


print(predict_labels)        

print( correct_labels.values )      # to get the correct labels in array form    
```

```python

count = 0

for i in range( 0 , len(correct_labels)):

    if predict_labels[i] == correct_labels[i]:
        count = count + 1
```

```python

accuracy = (count / len(correct_labels) ) * 100      # len( predict_labels ) 

accuracy
```

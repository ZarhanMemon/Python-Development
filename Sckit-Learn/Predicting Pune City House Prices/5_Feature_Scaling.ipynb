```python
import pandas as pd
import numpy as np
 
```

```python
df = pd.read_csv('california_housing_data.csv')

df[ 'income_categories'] = pd.cut( df['median_income'] , bins=[ 0 ,2 ,4, 6 , 8, np.inf],
                                                      labels=[ 1 ,2 , 3, 4, 5 ] )
```

```python
from sklearn.model_selection import StratifiedShuffleSplit

splitting = StratifiedShuffleSplit( n_splits = 1 , test_size = 0.2 , random_state = 42)
for train_split ,  test_split in splitting.split( df , df['income_categories'] ):
    train_data = df.loc[ train_split]
    test_data = df.loc[ test_split]

train_data ,test_data
```

```python

train_data.drop( columns=['income_categories'] , inplace = True)  

# inplace = True means that the changes will be made in the original dataframe and not a copy of it

test_data.drop( columns=['income_categories'] , inplace = True)
```

## Feature scaling in sklearn

Feature scaling is a crucial preprocessing step. Most machine learning algorithms perform poorly when input features have vastly different scales.

- In the California housing dataset, for example:

1. total_rooms ranges from 6 to over 39,000
2. median_income ranges from 0 to 15


If you don’t scale these features, models will give more importance to total_rooms simply because it has larger values.

### Why Scaling Is Needed
- Many models (like Linear Regression, KNN, SVMs, Gradient Descent-based algorithms) assume features are on a similar scale.
- Without scaling, features with larger ranges can dominate model behavior.
- Scaling makes training more stable and faster.



------

### 1. Min-Max Scaling (Normalization)

This method rescales the data to a specific range, usually [0, 1] or [-1, 1].

- it dont change/disrupt values , instead modulate according to the [min ,max] value.

Formula:

` scaled_value = (x - min) / (max - min)`

Use Scikit-Learn’s MinMaxScaler:

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler( feature_range=(0,1))  # scaling the data into range of 0 to 1

housing_min_max_scaled = scaler.fit_transform(housing)


housing_min_max_scaled
```

```python
# Convert the scaled data back into a DataFrame with the original column names and index
housing_minmax_scaled_df = pd.DataFrame(housing_min_max_scaled, 
                                        columns=housing.columns, index=housing.index)

housing_minmax_scaled_df
```

### Use feature_range=(0, 1) for models like neural networks.
### Sensitive to outliers — extreme values can distort the scale.

-------

## 2. Standardization (Z-score Scaling)

This method centers the data around 0 and scales it based on standard deviation.

Formula:

    ` standardized_value = (x - mean) / std

- Use Scikit-Learn’s StandardScaler:

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()  # scaling the data to have mean=0 and std=1

housing_std_scaled = scaler.fit_transform(housing)


housing_std_scaled
```

```python
# Convert the scaled data back into a DataFrame with the original column names and index
housing_std_scaled_df = pd.DataFrame(housing_std_scaled, 
                                     columns=housing.columns, index=housing.index)

housing_std_scaled_df
```

- Resulting features have zero mean and unit variance
- Robust to outliers compared to min-max scaling
- Recommended for most ML algorithms, especially when using gradient descent

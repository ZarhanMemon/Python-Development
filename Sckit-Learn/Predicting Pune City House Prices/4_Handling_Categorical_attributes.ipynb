```python
import pandas as pd
import numpy as np
 
```

```python
df = pd.read_csv('california_housing_data.csv')

df[ 'income_categories'] = pd.cut( df['median_income'] , bins=[ 0 ,2 ,4, 6 , 8, np.inf],
                                                      labels=[ 1 ,2 , 3, 4, 5 ] )
```

## Handling Categorical values

Most machine learning algorithms work best with numerical data. But real-world datasets often contain categorical or text attributes. Let’s understand how to handle these in Scikit-Learn using the ocean_proximity column from the California housing dataset as an example.

-----

### -> Categorical Attributes

Text columns like "ocean_proximity" are not free-form text but limited to a fixed set of values (e.g., "NEAR BAY", "INLAND"). These are known as categorical attributes.

- Example:

```python
housing_cat = housing[["ocean_proximity"]]

set(housing_cat["ocean_proximity"])  # to see the unique values in the ocean_proximity column

```

-----

## 1. Ordinal Encoding

Scikit-Learn's OrdinalEncoder can convert categories to numbers:

```python
from sklearn.preprocessing import OrdinalEncoder # string categorical data to numerical data 
                                                  # ( A = 0 , B = 1 , C = 2 , D = 3 , E = 4 ) 

ordinal_encode = OrdinalEncoder()

housing_cat_ord_encoded = ordinal_encode.fit_transform(housing_cat)

# fit_transform() method is used to fit the encoder to the 
# data and transform the data into encoded form in one step.
```

This will output a 2D NumPy array with numerical category codes.

- To see the mapping:

```python
ordinal_encode.categories_
# Output: array(['<1H OCEAN', 'INLAND', 'ISLAND', 'NEAR BAY', 'NEAR OCEAN'])

#------

housing_cat_ord_encoded   # numpy arr contain num data for refering catogory data
```

### ⚠️ Caution: Ordinal encoding implies an order between categories, which may not be true here. For example, it treats INLAND (1) as closer to <1H OCEAN (0) than NEAR OCEAN (4), which might not make sense.

------

## 2. One-Hot Encoding

For unordered categories, one-hot encoding is a better choice. It creates one binary column per category.

```python
from sklearn.preprocessing import OneHotEncoder  
# to convert the categorical data into multipe cols acc to no of catg contain
# [0,1] values in each col for each catg if it exist in the data or not


onehot_encoder = OneHotEncoder()

# unknown_

housing_cat_oh_encoded = onehot_encoder.fit_transform(housing_cat)

onehot_encoder.categories_
```

This gives a sparse matrix (efficient storage for mostly zeros).

- To convert it to a regular NumPy array:

```python
housing_cat_oh_encoded.toarray()
```

```python
# Convert the one-hot encoded data into a DataFrame

housing_cat_oh_encoded_df = pd.DataFrame(housing_cat_oh_encoded.toarray(), 
                                         columns=['<1H OCEAN', 'INLAND', 'ISLAND', 'NEAR BAY', 'NEAR OCEAN'], 
                                         index=housing_features.index) 


housing_cat_oh_encoded_df
```

```python

housing = pd.concat([housing.drop(columns=['ocean_proximity']), housing_cat_oh_encoded_df], axis=1)  
# concating the label and one hot encoded data into one dataframe


housing
```

------

4. Summary

- Method	                Use When	        Output Type
1. OrdinalEncoder	=>    Categories have an order	=> 2D NumPy array
2. OneHotEncoder	  =>   Categories are unordered	=> Sparse or dense


Using the right encoding ensures your model learns correctly from categorical features.

## Random Shuffle and Split : Creating a Test Set

```python


import numpy as np
```

```python


def shuffle_split_test ( dataframe , test_ratio ):

    np.random.seed(42)          # Set the seed for reproducibility - return same shuffled df output

    shuffled_data_indices = np.random.permutation( dataframe.index )   # This returns shuffled indices of df

    test_data_size = int( len(dataframe) * test_ratio )  # Calculate the number of samples for the test set
     
    test_data_indx  = shuffled_data_indices[ : test_data_size ]   # Select the first 'test_data_size' indices  for the test set
    train_data_indx = shuffled_data_indices[ test_data_size : ]   # Select the remaining indices for the training set

    return dataframe.iloc[train_data_indx], dataframe.iloc[test_data_indx]  # Return the training and test sets based on the shuffled indices from dataframe
```

```python

train , test = shuffle_split_test( df , 0.2 )
```

## Problem with Random Test Set Extraction
**Using pure random shuffling to create test sets can lead to sampling bias**, especially with the California Housing dataset:


# 1. Missing Categories in Training Set — 
- **In extreme cases, all samples of a particular category (e.g., "<1H OCEAN") may randomly end up in the test set, leaving the training set with **zero examples** of that category. This causes the model to fail when it encounters such samples during evaluation.**

**Solution**: Use **stratified sampling** to ensure the test set maintains the same distribution as the original dataset for important categorical features like `ocean_proximity`.
